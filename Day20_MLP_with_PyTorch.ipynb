#**Creating Linear Layers using PyTorch**

## **Let's generate a random input for our linear layer and plug it into our layer**

```python
# Step 1: let's generate 4 random samples of (x1, x2) for the above linear network
torch.manual_seed(0)                    # for reproducibility (you will get the same random number every time you run this cell)
number_of_samples     = 1
random_input          = torch.randn(number_of_samples, 2) # we have 4 samples each has two predictors (x1 and x2)
print(f'input numbers: \n{random_input.numpy()}\n')
```

## **Let's build a linear layer**

```python
torch.manual_seed(1)                    # for reproducibility
# construction of a linear layer
num_neuron_input    = 2
num_neuron_output   = 3
input_layer_1 = nn.Linear( num_neuron_input, num_neuron_output)
```

##**Inspecting the weights of a linear layer**

```python
# Print the weights of the linear layer
print(f'Weights: \n{input_layer_1.weight.data}') # matrix of size: [ out_feature by in_feature ]
# Print the biases of the linear layer (if they exist)
print(f'Biases: \n{input_layer_1.bias.data}')    # vector of size: [ out_feature ]
```

## **Let's build the next linear layer and visualize the weights and bias parameters**

```python
torch.manual_seed(1)                    # for reproducibility
num_neuron_input    = 3
num_neuron_output   = 1
output_layer_1      = nn.Linear(num_neuron_input, num_neuron_output)
```

```python
# print the weights of the linear layer
print(f'Weights: \n{output_layer_1.weight.data}') # matrix of size: [ out_feature by in_feature ]
# print the biases of the linear layer (if they exist)
print(f'Biases: \n{output_layer_1.bias.data}')    # vector of size: [ out_feature ]
```

##**Apply forward pass through the first layer of the MLP**

```python
# Step 2: apply forward pass through the network
output                = input_layer_1(random_input)
print(f'output layer value: \n{output.data.numpy()}\n')

# attach a Sigmoid() activation function
activation_function = nn.Sigmoid()
output                = activation_function(output)
print(f'output layer value after sigmoid activation function: \n{output.data.numpy()}\n')
```

##**Apply forward pass through the second layer of the MLP to the final output**

```python
# Step 2: apply forward pass through the network
output                = output_layer_1(output)
print(f'output layer value: \n{output.data.numpy()}\n')

# attach a Sigmoid() activation function
activation_function = nn.Sigmoid()
output                = activation_function(output)
print(f'output layer value after sigmoid activation function: \n{output.data.numpy()}\n')
```
