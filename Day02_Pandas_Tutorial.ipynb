## Selecting Rows in DataFrames using `loc` and `iloc`:
Simply put:
- `loc` gets DataFrame rows and columns by __labels/names__
- `iloc` gets DataFrame rows and columns by __index/position__

###Selecting rows in a DataFrame
`loc` gets DataFrame rows and columns by __labels/names__


```python
print(df_titanic.loc[0])   # 1 is really a "name", not a number
```

```python
print(df_titanic.loc[880])   # 880 is really a "name", not a number
```

Let's take a subset of titanic and try to use `loc` and `iloc`:

```python
subset = df_titanic.loc[1:6] # since it's a label, it will take rows labeled 1, 2, 3, 4, 5, and 6.
print(subset)
```

```python
subset = df_titanic.loc[800:805] # since it's a label, it will take rows labeled 800, 801, 802, 803, 804, and 805.
print(subset)
```

```python
subset.loc[800] # will show the 1st row in the DataFrame called 'subset'
```

```python
subset.loc[805] # will show the 6th row in the DataFrame called 'subset'
```

```python
subset.loc[806] # DataFrame called 'subset' has only 6 rows, hence ERROR
```

###Selecting Rows in a DataFrame
`iloc` gets DataFrame rows and columns by __index/position__

```python
subset.iloc[5]  #works
```

Why might we want a subset of the columns of a DataFrame?

Sometimes you don't need all of the columns and just want to work with a __subset__ of the colummns of the original dataset. Other times, you may want to reorder the columns in your dataset.

Here's how you would do either of those:
The syntax for subsetting columns from a DataFrame (`df`) is:
- One column: `df['column_name']`
- Multiple columns: `df[['column1', 'column2', 'target']]`

So, if we wanted to look at the `price` column, we could do:

```python
prices = df_rest['price']
prices.head()
```

## Column Subset Example:

Imagine you want to only work with `'rain'`, `'hun'`, and `'target'` from the 'restaurant' dataset:


```python
prices = df_rest[ ['rain', 'hun', 'target'] ]
prices.head()
```

Maybe you want to re-order your new subset so that `rain` and `hun` are switched:

```python
prices_reordered = df_rest[ ['hun', 'rain', 'target'] ]
prices_reordered.head()
```
